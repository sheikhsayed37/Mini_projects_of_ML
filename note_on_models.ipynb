# ১. Logistic Regression (LR)

## কী?

Supervised Classification Algorithm।

Binary Classification (0/1, Yes/No)-এর জন্য ব্যবহার হয়।

## কীভাবে কাজ করে?

```
Input Feature
      ↓
Linear Equation
      ↓
Sigmoid Function
      ↓
Probability (0-1)
      ↓
Class Prediction
```

Probability ≥ 0.5 → Class 1

Probability < 0.5 → Class 0

## কোথায় ব্যবহার হয়?

* Diabetes Prediction
* Titanic Survival
* Heart Disease
* Spam Detection

## সুবিধা

✔ Fast

✔ Simple

✔ Explainable

## অসুবিধা

✘ Complex Pattern ধরতে পারে না

## Scaling

✅ লাগে

## Parameters

### max_iter

কতবার Train করবে

```
max_iter=500
```

---

### penalty

Regularization

```
l1
l2
```

Overfitting কমায়।

---

### C

Regularization Strength

ছোট C → বেশি Regularization

বড় C → কম Regularization

---

### solver

Optimization Method

```
lbfgs
liblinear
```

## ⭐ একটি গুরুত্বপূর্ণ বিষয়

**Parameter কেন ব্যবহার করি?**

* **`n_estimators`** → কয়টি Tree তৈরি হবে।
* **`max_depth`** → Model বেশি complex হবে নাকি কম হবে।
* **`min_samples_split`** → খুব ছোট group-কে split করা বন্ধ করতে।
* **`C` (LR/SVM)** → Overfitting নিয়ন্ত্রণ করতে।
* **`n_neighbors` (KNN)** → কয়টি প্রতিবেশী দেখে সিদ্ধান্ত নেবে।
* **`kernel` (SVM)** → Linear নাকি Non-linear boundary ব্যবহার করবে।
* **`n_components` (PCA)** → কতটি feature রেখে dimension কমাবে।
* **`n_clusters` (KMeans)** → কতটি cluster বানাবে।
* **`eps` (DBSCAN)** → একটি point-এর আশেপাশে কত দূর পর্যন্ত প্রতিবেশী ধরা হবে।
